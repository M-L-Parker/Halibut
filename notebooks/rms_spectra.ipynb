```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as pl
```

```python

fig=pl.figure(figsize=(12,8))
ax1=pl.subplot(211)
ax1.set_xscale('log')
ax1.set_xlim(2,10)


ebins=np.logspace(np.log10(0.3),np.log10(10),1001)
mean_es=[(x+y)/2. for x, y in zip(ebins[:-1],ebins[1:])]
densities=[1e-08,3e-08,1e-07,3e-07]

for density in densities:
    indata=np.loadtxt('rms_spectrum_density_%s.dat' % str(density))
    pl.plot(mean_es,indata,lw=1,label=str(density))
pl.legend(loc='best',frameon=False)
    
    
ax2=pl.subplot(212)
ax2.set_xscale('log')
ax2.set_xlim(0.8,10)

for density in densities:
    indata=np.loadtxt('rms_spectrum_density_%s.dat' % str(density))
    pl.plot(mean_es,indata/np.loadtxt('rms_spectrum_density_%s.dat' % str(1e-8)),lw=1,label=str(density))
    
pl.savefig('fvar_density.pdf',bbox_inches='tight')
```
