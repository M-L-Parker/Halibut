```python
%matplotlib inline
import numpy as np
from matplotlib import pyplot as pl
from glob import glob
```

```python
spectra=glob('av_covar_density_*')
densities=[float(x.split('_')[-1].split('.')[0]) for x in spectra]
spectra=[y for x,y in sorted(zip(densities,spectra))]

colours=['dodgerblue','forestgreen','goldenrod','orangered','red'][::-1]

n_energies=100

fig=pl.figure(figsize=(10,6))

ax=pl.subplot(111)
ax.set_xscale('log')
ax.set_xlim(0.5,10)
# ax.set_ylim(0,0.65)

energy_bins=np.logspace(np.log10(0.3),np.log10(10),n_energies)

for i,filename in enumerate(spectra):
    density=float(filename.split('_')[-1].split('.')[0])
#     print density
#     print filename
    temp_data=np.loadtxt(filename)
    pl.plot(energy_bins,temp_data,label=density,color=colours[i],lw=1)

pl.legend(loc='best',frameon=False)
```
